```python
# df_pyspark.withColumn('Experience After 2 Years', df_pyspark['Experience']+2) not an in place operation have to assign
# to new variable
df_pyspark = df_pyspark.withColumn('Experience After 2 Years', df_pyspark['Experience']+2)
```

```python
df_pyspark.show()
```

```python
# Drop the columns
# df_pyspark.drop('Experience After 2 Years') this is not inplace operation
df_pyspark = df_pyspark.drop('Experience After 2 Years')
```

```python
df_pyspark.show()
```

```python
# Rename the columns
df_pyspark.withColumnRenamed('Name','New Name').show()
```
